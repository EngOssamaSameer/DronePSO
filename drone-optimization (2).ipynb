**Particle Swarm Optimization (PSO)**

In this problem, we want to find the best drone design that gives good performance. The main goal is to reduce the drag during flight and minimize the wobbling caused by wind. Since these two factors affect the drone’s stability and efficiency, we use Particle Swarm Optimization (PSO) to search for a better solution. Each particle represents a possible solution, and the swarm gradually moves toward the best solution based on the fitness of each particle.

```python
from random import uniform
import math
import random
```

```python
class Particle:
    def __init__(self,x,y,inertia,cognitive_constant,social_constant):
        self.x = x
        self.y = y
        self.fitness = float("inf")
        self.velocity = 0
        self.best_x = x
        self.best_y = y
        self.best_fitness = float("inf")
        self.inertia = inertia
        self.cognitive_constant = cognitive_constant
        self.social_constant = social_constant
        
```

```python
INERTIA = 0.2
COGNITIVE_CONSTANT = 0.35
SOCIAL_CONSTANT = 0.45
```

```python
#this function to generate number of swarm 
def generate_swarm(number_of_swarm):
    particles = []
    for particle in range(number_of_swarm):
        particle  = Particle(uniform(-10,10),uniform(-10,10),INERTIA,
                             COGNITIVE_CONSTANT,SOCIAL_CONSTANT)
        particles.append(particle)
    return particles
```

```python
# here we was created a swarm consist of 4 praticle
swarm = generate_swarm(4)
```

```python
# lets see the swarm
def print_swarm(swarm):
    for i, particle in enumerate(swarm):
        print(f"Particle {i+1}:")
        print(f"  x = {particle.x}")
        print(f"  y = {particle.y}")
        print(f"  fitness = {particle.fitness}")
        print(f"  velocity = {particle.velocity}")
        print(f"  best_x = {particle.best_x}")
        print(f"  best_y = {particle.best_y}")
        print(f"  best_fitness = {particle.best_fitness}")
        print()
```

```python
print_swarm(swarm)
```

```python
# function to calculate fitness basedon given roules
def calculate_fitness(x,y):
    return ((x+2*y-7)**2) + ((2*x+y-5)**2)
```

```python
#funcion to update fitness and set best fitness and best x and y
def update_fitness(particle):
    particle.fitness = calculate_fitness(particle.x,particle.y)
    if particle.fitness < particle.best_fitness :
        particle.best_fitness = particle.fitness 
        particle.best_x = particle.x
        particle.best_y = particle.y
```

```python
# functionto get best fitness and best particle
def get_best(swarm):
    best_fitness = float("inf")
    best_particle = None
    
    for particle in swarm:
        update_fitness(particle) 
        
        if particle.fitness < best_fitness :
            best_fitness = particle.fitness
            best_particle = particle
            
    return best_particle
```

```python
# Calculate Inertia
def calculate_inertia(inertia_constant,velocity):
    return inertia_constant*velocity
```

```python
# Calculate Cognative Acceleation
def calculate_cognitive_acceleration(cognaitive_constan):
    return cognaitive_constan*uniform(0,1)
```

```python
print(calculate_cognitive_acceleration(COGNITIVE_CONSTANT))
```

```python
# Calculate Social Acceleation
def calculate_social_acceleration(social_constan):
    return social_constan*uniform(0,1)
```

```python
print(calculate_social_acceleration(SOCIAL_CONSTANT))
```

```python
# Calculate Distance
def calculate_distance(best_x,best_y,current_x,current_y):
    return math.sqrt((best_x - current_x)**2 +(best_y - current_y)**2)
```

> The cognitive component is calculated by finding the cognitive acceleration, using the 
function that we defined in an earlier section, and the distance between the particle’s 
best position and its current position:

```python
#Calculate Conitive
def calculate_cognitive(cognitive_constant,particle_best_x,
                       particle_best_y,particle_current_x,
                       particle_current_y):
    acceleration = calculate_cognitive_acceleration(COGNITIVE_CONSTANT)
    disance = calculate_distance(particle_best_x,
                                 particle_best_y,
                                 particle_current_x,
                                 particle_current_y)
    return acceleration * disance
    
```


> The social component is calculated by finding the social acceleration, using the function that we defined earlier, and the distance between the swarm’s best position and the particle’s current position:



```python
#Calculate Social
def calculate_social(csocial_constant,swarm_best_x,
                       swarm_best_y,particle_current_x,
                       particle_current_y):
    acceleration = calculate_social_acceleration(SOCIAL_CONSTANT)
    disance = calculate_distance(swarm_best_x,
                                 swarm_best_y,
                                 particle_current_x,
                                 particle_current_y)
    return acceleration * disance
    
```

```python
print(calculate_social(
    SOCIAL_CONSTANT,
    1,
    3,
    4,
    5
))
```

> The update function wraps everything that we have defined to carry out the actual 
update of a particle’s velocity and position. The velocity is calculated by using the inertia 
component, cognitive component, and social component. The position is calculated by 
adding the new velocity to the particle’s current position:


```python

def update_particle(cognitive_constant,csocial_constant,
                    particle_velocity,particle_best_x,
                    particle_best_y,swarm_best_x,swarm_best_y,
                    particle_current_x,particle_current_y):
    inertia = calculate_inertia(INERTIA,particle_velocity)
    cognitive = calculate_cognitive(cognitive_constant,particle_best_x,
                    particle_best_y,particle_current_x,
                    particle_current_y)
    social = calculate_social(csocial_constant,swarm_best_x,
                       swarm_best_y,particle_current_x,
                       particle_current_y)
    
    velocity = inertia + cognitive + social

    new_x = particle_current_x + velocity
    new_y = particle_current_y + velocity

    return new_x, new_y, velocity
```

```python
# Train Swarm
best_particle = None
for i in range(100):
    best_particle = get_best(swarm)
    for particle in swarm:
       x, y, velocity = update_particle(COGNITIVE_CONSTANT,
                    SOCIAL_CONSTANT,particle.velocity,
                    particle.best_x,particle.best_y,
                    best_particle.best_x,best_particle.best_y,
                    particle.x,particle.y)
       
       particle.x = x
       particle.y = y
       particle.velocity = velocity
       update_fitness(particle)
    
```
